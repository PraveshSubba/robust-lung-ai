# Phase 2–3 — Consolidated Experiment Results

**Project:** RobustLungAI — Robust Respiratory Sound Classification under Noise and Domain Shift using Audio Spectrogram Transformers.

## Objective

Consolidate the recorded baseline, AST fine-tuning, focal-loss, supervised-contrastive-learning and augmentation-ablation results into a single table for analysis and documentation.

## Source experiments

* Phase 2: classical and CNN baselines.
* Phase 3A: AST with cross-entropy.
* Phase 3B: AST with focal loss.
* Phase 3C: AST with focal loss and supervised contrastive learning.
* Phase 3D: SpecAugment and Mixup ablation.

## Validation requirements

Before combining results, verify that the source CSVs use compatible column names, metric units and split labels. Validation and test results must remain distinguishable.

## Scope

This notebook aggregates recorded metrics; it does not retrain models or establish which configuration generalizes best. Phase 4 robustness results are maintained separately unless explicitly added with compatible metrics and documented evaluation protocols.


```python
import pandas as pd

phase2 = pd.read_csv(
    "/kaggle/input/notebooks/praveshsubba/02-baseline-models/results.csv"
)

phase3 = pd.read_csv(
    "/kaggle/input/notebooks/praveshsubba/03a-ast-finetuning-ce/results_phase3.csv"
)

phase3b = pd.read_csv(
    "/kaggle/input/notebooks/praveshsubba/03b-ast-focal-loss/results_phase3b.csv"
)

phase3c = pd.read_csv(
    "/kaggle/input/notebooks/praveshsubba/03c-ast-subcon/results_phase3c.csv"
)

phase3d = pd.read_csv(
    "/kaggle/input/notebooks/praveshsubba/03d-ast-augmentation-ablation/results_phase3d.csv"
)
```

```python
final_results = pd.concat(
    [
        phase2,
        phase3,
        phase3b,
        phase3c,
        phase3d
    ],
    ignore_index=True
)

display(final_results)
```

```python
final_results.to_csv(
    "/kaggle/working/results_phase3_combined.csv",
    index=False
)
```

## Output and Interpretation

The exported CSV consolidates the recorded baseline and Phase 3 experiment metrics. Each row retains its source experiment and evaluation split.

The combined table is intended for reporting and visualization. It should not be interpreted as a direct ranking of models unless the data splits, metric definitions, preprocessing and evaluation protocols are compatible.

Phase 4 robustness results—including noise, device and cross-dataset experiments—should be added in a separate aggregation step after their output schemas and evaluation protocols have been verified.
